```python
from sklearn.preprocessing import StandardScaler
X = df.drop('CUST_ID', axis=1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
print(X_scaled_df)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

cgrps = []

for i in range(1, 11):
    Kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    Kmeans.fit(X_scaled)
    cgrps.append(Kmeans.inertia_)
    
plt.figure(figsize=(10, 5))
plt.plot(range(1, 11), cgrps, marker='o', linestyle='--')
plt.title("The Elbow Method")
plt.xlabel('Number of cluster (K)')
plt.ylabel('crgps (Error Score)')
plt.grid(True)
plt.show()
```

```python
# Initiate Kmeans with n_clusters.abs
Kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)

clusters = Kmeans.fit_predict(X_scaled)

df["cluster_group"] = clusters

df[['CUST_ID', 'cluster_group']].head()
```

```python
# Analyzing the clusters
cluster_analysis = df.drop('CUST_ID', axis=1).groupby('cluster_group').mean()

important_columns = ['BALANCE', 'PURCHASES', 'CASH_ADVANCE', 'CREDIT_LIMIT', 'PAYMENTS', 'MINIMUM_PAYMENTS']

print(cluster_analysis[important_columns])
```
